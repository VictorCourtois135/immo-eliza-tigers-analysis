----------------------------------------------------------------------------------------------------------

**Deleting rows with Nan prices**  
Rows without a price are useless

```python
df_immo = df_immo.dropna(subset=['price'])
```

**Substituting sentinel values with NaN**  
this to keep the other infos in the rows

```python
df_immo.loc[df_immo['floor_number'] > 100, 'floor_number'] = pd.NA
```

--------------------------------------------------------------------------------------------------------------

alligning native data types

```python
# mapping datatypes
type_map = {
    'price': 'float64',
    'bedrooms': 'Int64',
    'postal_code': 'Int64'
}

# vectorial conversion on the columns
df_immo = df_immo.astype(type_map)
```

**checking on destructive values ​​of scale**

```python
df_immo.describe(include='all')
```

filtering out not plausible values in *price* and *living_area_m2*

```python
# price
cond_low_price = (df_immo['price'] <= 1)
df_immo = df_immo[~ cond_low_price]

# livingt_area_m2
cond_low_area = (df_immo['living_area_m2'] <= 1)
df_immo = df_immo[~ cond_low_area]
```
