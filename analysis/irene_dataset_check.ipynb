```python
columns_to_control = ['living_area_m2', 'total_area_m2','price']
suspect_vals_cols = ['bathrooms', 'bedrooms', 'floor_number']
for suspect in suspect_vals_cols:
    if suspect == 'bathrooms':
        cond = (df_immo['bathrooms'] > 5.00)
    elif suspect == 'bedrooms':
        cond = (df_immo['bedrooms'] > 5.00)
    else:
        cond = (df_immo['floor_number'] == 999.00)

    print(f"Property url: {df_immo.loc[cond, 'property_url']}")
```

```python
# printing temporary df

condition = (df_immo['bathrooms'] > 5.00)
df_immo_target = df_immo[condition].sort_values(by='bathrooms', ascending=False)

indexes_to_be_deleted = df_immo_target[df_immo_target['price'].isna()].index.tolist()
print(indexes_to_be_deleted)

df_immo_to_print = df_immo_target[df_immo_target['price'].notna()]

df_immo_to_print.loc[condition, ['bathrooms'] + columns_to_control]
```

```python

# printing values in columns_to_control for each row and deleting rows with NaN price

for index, row in df_immo_to_print.iterrows():
    print(f"property with {row['bathrooms']} in 'bathrooms' column (Index: {index}) has:\n")
    for column in columns_to_control:
        print(f"  {row[column]} in {column}.\n")

# deletion of found rows with NaN price value
df_immo.drop(indexes_to_be_deleted)
deleted_indexes = ", ".join(map(str, indexes_to_be_deleted))
print(f'Rows with labels {deleted_indexes} with NaN price habe been successfully deleted.')
```

# **Investigate NaN data**

```python
df_immov = df_immovlan.copy()

# printing proportion of missing data in decreasing order of percentage
# print((df_immov.isna().mean()* 100).sort_values(ascending=False))

# printing the columns names with highest values of nan %
filtrated_col = (
    df_immov.isna().mean()* 100).sort_values(ascending=False)[
    (df_immov.isna().mean()* 100).sort_values(ascending=False) > 2]
print("Columns with nan percentage higher than 2:")
print(filtrated_col)

```

```python
# grouping data by "property_type" and checking again for nan %
grouped_nan_perc = df_immov['floor_number'].isna().groupby(df_immov['property_type']).mean() * 100
print(grouped_nan_perc)
print("-" * 30)
print("\n")
print(f'⚠️ "Apartment" type has {grouped_nan_perc.loc['Apartment'].round(2)} percent nan.')
print("Fill 'floor_number' with median or mode.")
print("\n")
print(f'⚠️ "House" type has {grouped_nan_perc.loc['Apartment'].round(2)} percent nan.')
print("No 'floor_number' data for this group.")
```
