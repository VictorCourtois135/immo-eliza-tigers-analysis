```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
import pandas as pd

# --- 1. MOCK DATA SETUP ---
regions = ["Belgium", "Flanders", "Wallonia"]

cities_most = {
    "Belgium": ["eernegegem", "knokke-heist", "knokke", "seneffe", "schilde"],
    "Flanders": ["eernegegem", "knokke-heist", "knokke", "schilde", "schoten"],
    "Wallonia": ["seneffe", "pecq", "waimes", "libin", "warsage"],
}

prices_most = {
    "Belgium":,
    "Flanders":,
    "Wallonia":,
}

cities_least = {
    "Belgium": [
        "bailleux",
        "blaimont",
        "hastiere-par-dela",
        "wasmes",
        "marchienne-au-pont",
    ],
    "Flanders": ["wervik", "herselt", "westende", "rekkem", "oostrozebeke"],
    "Wallonia": [
        "bailleux",
        "blaimont",
        "hastiere-par-dela",
        "wasmes",
        "marchienne-au-pont",
    ],
}

prices_least = {
    "Belgium":,
    "Flanders":,
    "Wallonia":,
}

# --- 2. PALETTE & PLOT SETUP ---
color_most = "#E05A47"  # Tomato Red
color_least = "#4682B4"  # Steel Blue

fig, axes = plt.subplots(3, 2, figsize=(15, 17))  # Made slightly taller for numbers
fig.suptitle(
    "Most vs Least Expensive Municipalities by Region",
    fontsize=16,
    fontweight="bold",
    y=0.98,
)


# Custom formatter for European Currency style: €1.200.000 (No decimals)
def euro_formatter(x, pos=None):
    return f"€{int(x):,}".replace(",", ".")


formatter = ticker.FuncFormatter(euro_formatter)

# --- 3. LOOP AND GENERATE SUBPLOTS ---
for i, region in enumerate(regions):
    # Left Column: Most Expensive
    ax_most = axes[i, 0]
    bars_most = ax_most.bar(
        cities_most[region], prices_most[region], color=color_most
    )
    ax_most.set_title(f"Most Expensive in {region}", fontsize=12, pad=15)

    # ADD NUMERICAL LABELS ON BARS (MOST EXPENSIVE)
    ax_most.bar_label(
        bars_most,
        labels=[euro_formatter(val) for val in prices_most[region]],
        padding=4,
        fontsize=8.5,
        fontweight="bold",
    )

    # Right Column: Least Expensive
    ax_least = axes[i, 1]
    bars_least = ax_least.bar(
        cities_least[region], prices_least[region], color=color_least
    )
    ax_least.set_title(f"Least Expensive in {region}", fontsize=12, pad=15)

    # ADD NUMERICAL LABELS ON BARS (LEAST EXPENSIVE)
    ax_least.bar_label(
        bars_least,
        labels=[euro_formatter(val) for val in prices_least[region]],
        padding=4,
        fontsize=8.5,
        fontweight="bold",
    )

    # Apply global styling adjustments to both columns
    for ax in [ax_most, ax_least]:
        ax.set_ylabel("Average Price (€)", fontsize=10)
        ax.set_xlabel("City", fontsize=10, labelpad=8)

        # Fix overlapping text: Rotate city names and align them right
        ax.set_xticklabels(
            ax.get_xticklabels(), rotation=35, ha="right", fontsize=9
        )

        # Standardize Y-Axis format to European dots (€1.000.000)
        ax.get_yaxis().set_major_formatter(formatter)

        # Clean look: Remove top and right borders
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

        # Expand Y-limits slightly so the bar numbers don't touch the top box border
        ax.set_ylim(0, ax.get_ylim()[1] * 1.12)

        # Soft gridlines on the Y-axis for easier reading
        ax.grid(axis="y", linestyle="--", alpha=0.3)

# --- 4. ADJUST PADDING & SHOW ---
plt.tight_layout(rect=[0, 0, 1, 0.95], h_pad=5.0, w_pad=3.0)
plt.show()
```
