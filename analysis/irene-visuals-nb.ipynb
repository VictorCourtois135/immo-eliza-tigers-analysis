```python
# imports configuration
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io
from IPython.display import display, HTML

# forcing pandas to display whole url
pd.set_option('display.max_colwidth', None)

# pforcing pandas to display all the columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', 1000)

# path to final cleaned dataset in data/cleaned
CLEANED_CSV_PATH = "../data/cleaned/properties_final_irene.csv"
# path to plots directory
PLOTS_PATH = "../images/"

# reading the file
data_df = pd.read_csv(CLEANED_CSV_PATH)

# checking dataset
data_df.describe(include='all')
```

### **Q2**: What is the proportion of missing values per column?

```python
print("="*30 + " DATASET STRUCTURE " + "="*30)
print("\n")
print(f"Number of (motionless) observations: {data_df.shape[0]}\n")
print(f"Number of properties features (columns): {data_df.shape[1]}\n")
print("\n")

print("="*30 + " MISSING VALUES " + "="*30)
print("\n")
# calculation of the proportion of missing values
missing_prop = (data_df.isnull().sum() / len(data_df)) * 100
missing_prop = missing_prop[(missing_prop > 0)].sort_values(ascending=False)
print(f"Percentage of missing values for each column:")
print("\n")
display(missing_prop)
print("\n")

# data structure visual
plt.figure(figsize=(10,6))
structure_bar = sns.barplot(
    x=missing_prop.values, 
    y=missing_prop.index, 
    hue=missing_prop.index,
    palette="viridis",
    legend=False
)
plt.title(
    "Proportion of Missing Values per Column (%)",
    fontsize=14,
    pad=15
)
plt.xlabel(
    "Percentage (%)",
    fontsize=12
)
plt.ylabel(
    "Columns",
    fontsize=12
)
structure_bar.set_title("Dataset Strucutre Visual")
sns.despine()
plt.savefig(
    f"{PLOTS_PATH}structure_barplot.png", 
    bbox_inches='tight', 
    dpi=330
)
plt.show()
plt.close()
print(f"Barplot successfully saved in {PLOTS_PATH}structure_barplot.png")
```

Missing data may be **Random** (for instance, 'price') or **Structural** (for instance, 'garden_area_m2').
