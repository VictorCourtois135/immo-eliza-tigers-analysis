```python
import pandas as pd
import seaborn as sns
import matplotlib .pyplot as plt
```

```python
df = pd.read_csv("../data/cleaned/properties_final.csv")
df.head
```

```python
df.describe()
```

```python
# Price = 1
print("=== PRICE = 1 ===")
print(df[df['price'] == 1][['property_url', 'city', 'price', 'living_area_m2']])

# living_area_m2 < 10
print("\n=== LIVING AREA < 10m² ===")
print(df[df['living_area_m2'] < 10][['property_url', 'city', 'price', 'living_area_m2']])

# bedrooms > 30
print("\n=== BEDROOMS > 30 ===")
print(df[df['bedrooms'] > 30][['property_url', 'city', 'price', 'bedrooms']])

# longitude > 7
print("\n=== WRONG LONGITUDE ===")
print(df[df['longitude'] > 7][['property_url', 'city', 'latitude', 'longitude']])

# km_from_nearby_city > 350
print("\n=== KM FROM CITY > 350 ===")
print(df[df['km_from_nearby_city'] > 350][['property_url', 'city', 'km_from_nearby_city']])

# floor_number > 100
print("\n=== FLOOR NUMBER > 100 ===")
print(df[df['floor_number'] > 100][['property_url', 'city', 'floor_number']])

# building_year > 2026
print("\n=== BUILDING YEAR IN FUTURE ===")
print(df[df['building_year'] > 2026][['property_url', 'city', 'building_year']])
```

```python
# Check unique values for categorical columns
print("=== PROPERTY TYPE ===")
print(df['property_type'].value_counts(dropna=False))

print("\n=== PROPERTY SUBTYPE ===")
print(df['property_subtype'].value_counts(dropna=False))

print("\n=== PRICE TYPE ===")
print(df['price_type'].value_counts(dropna=False))

print("\n=== STATE OF BUILDING ===")
print(df['state_of_the_building'].value_counts(dropna=False))

print("\n=== KITCHEN EQUIPPED ===")
print(df['kitchen_equipped'].value_counts(dropna=False))

print("\n=== EPC SCORE ===")
print(df['epc_score'].value_counts(dropna=False))

print("\n=== REGION ===")
print(df['region'].value_counts(dropna=False))
```
